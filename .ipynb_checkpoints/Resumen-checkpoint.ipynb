### Práctica 5 - Suavizado y realce
1. Añadir ruido a una imagen
```
<imagen_salida> = imnoise (<imagen_entrada>,’<tipo de ruido>’,<param1>,…)
```
2. Filtros espaciales
```
<filtro> = fspecial (‘<tipo>, [ <tamañoX> ,<tamañoY> ] , <param1>, …)
```
3. Suavizado en el dominio espacial
```
im_media = conv2(im, ones(11)/121; 'same');
im_mediana = medfilt2(im);
```
4. Suavizado en el dominio frecuencial
```
[tamx,tamy] = size(im);
[xx,yy] = meshgrid(1:tamx,1:tamy);
dist_centro = sqrt((xx-tamx/2).^2+(yy-tamy/2).^2);
ideal_lpf = (dist_centro < dist_corte)';
butte_lpf = (1+(dist_centro/dist_corte).^(2*n)))';
```
5. Realce
```
<suma_acumulada> = cumsum (<histograma_normalizado>);
<lut_igualacion> = uint8 (<suma_acumulada>*255);
<imagen_igualada> = histeq (<imagen_entrada,256);
<imagen_espefificada> = histeq(<imagen_entrada>, <histograma_especificado>);
```
